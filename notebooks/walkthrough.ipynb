# FoodHub delivery operations

A descriptive case study of food delivery demand, delivery time, ratings, and an illustrative commission rule. The original course CSV is not bundled; use an authorized copy in `../data/foodhub_order.csv`. This notebook imports tested functions rather than duplicating the analysis.

```python
from pathlib import Path
from foodhub_analysis.analysis import load_orders, analyze_orders
from foodhub_analysis.charts import save_charts

DATA_PATH = Path("../data/foodhub_order.csv")
if DATA_PATH.exists():
    orders = load_orders(DATA_PATH)
else:
    orders = None
    print("Place an authorized CSV at", DATA_PATH)
```

## Aggregate metrics

The loader checks required columns, duplicate order IDs, nonnegative time and cost, and expected category values. `Not given` is a rating category.

```python
if orders is not None:
    summary = analyze_orders(orders)
    for name, value in summary.items():
        print(name, value)
```

## Charts

Charts use aggregate counts and means and are saved in ignored `results/`. The source dataset contains no exact timestamps or driver-capacity information, so the weekday difference is descriptive.

```python
if orders is not None:
    for chart in save_charts(orders, Path("../results/charts")):
        print(chart)
```
